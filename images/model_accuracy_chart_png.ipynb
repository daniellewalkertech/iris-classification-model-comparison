```python
import matplotlib.pyplot as plt

# Model names
models = [
    "Decision Tree",
    "Rule-Based",
    "Naive Bayes",
    "Logistic Regression",
    "KNN",
    "SVM"
]

# Accuracy values
accuracies = [96.67, 96.67, 96.67, 93.33, 93.33, 96.67]

# Create bar chart
plt.figure(figsize=(10, 6))
plt.bar(models, accuracies)

# Labels and title
plt.title("Iris Classification Model Accuracy Comparison")
plt.xlabel("Models")
plt.ylabel("Accuracy (%)")
plt.ylim(90, 100)
plt.xticks(rotation=20)
plt.tight_layout()

# Save image
plt.savefig("model-accuracy-chart.png", dpi=300, bbox_inches="tight")

# Show chart
plt.show()
```
